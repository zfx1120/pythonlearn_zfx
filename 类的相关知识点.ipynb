#类的相关知识点

```python
class Student:
    school="我是一个学校的名字" # 类属性(类变量)
    def __init__(self,name,age):
        self.name=name# 实例属性(实例变量)
        self.age=age
 
stu1 = Student('张三', age=28)
stu2 = Student('李四', 32)
 
""" 调用实例属性: 只能用实例对象调用, 不能用类对象调用 """
print(stu1.name)
print(stu2.name)
 
print(getattr(stu1, 'age'))
print(getattr(stu1, 'adres', '该实例属性不存在'))
"""
调用类属性: 既可以用类对象调用(推荐), 也可以用实例对象调用
注意: 当实例属性和类属性同名时, 实例对象优先调用实例属性
"""
 
 
print(Student.school)
print(stu1.school)# 本质: type(stu1).school
print(stu2.school) # 本质: type(stu2).school
print(getattr(Student, 'school'))
print(getattr(Student, 'adres', '该类属性不存在'))




```

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand  # 实例属性
        self.model = model  # 实例属性
 
# 创建 Car 实例
my_car = Car("Toyota", "Corolla")
 
# 输出属性值
print(my_car.brand)  # 输出: Toyota
print(my_car.model)  # 输出: Corolla
 
# 删除 model 属性
delattr(my_car, 'model')
 
# 尝试访问已删除的属性
try:
    print(my_car.model)  # 会引发 AttributeError
except AttributeError:
    print("model 属性已被删除")  # 输出: model 属性已被删除
```
